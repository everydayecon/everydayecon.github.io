```python
import pandas as pd

whl = pd.read_csv("whl_2025 - whl_2025.csv")
matchups = pd.read_csv("WHSDSC_Rnd1_matchups - matchups.csv")

whl.head()
```

```python
# Aggregate to one row per game
games = (
    whl
    .groupby(["game_id","home_team","away_team"], as_index=False)
    .agg({
        "home_goals":"sum",
        "away_goals":"sum",
        "home_shots":"sum",
        "away_shots":"sum",
        "home_xg":"sum",
        "away_xg":"sum"
    })
)

games["home_win"] = (games["home_goals"] > games["away_goals"]).astype(int)

games.head()
```

```python
import numpy as np

# Create team results
home_stats = games[["home_team","home_goals","away_goals","home_win"]].copy()
home_stats.columns = ["team","goals_for","goals_against","win"]

away_stats = games[["away_team","away_goals","home_goals","home_win"]].copy()
away_stats.columns = ["team","goals_for","goals_against","home_win"]
away_stats["win"] = 1 - away_stats["home_win"]
away_stats = away_stats.drop(columns=["home_win"])

team_games = pd.concat([home_stats, away_stats])

league_table = (
    team_games
    .groupby("team")
    .agg(
        games_played=("win","count"),
        wins=("win","sum"),
        goals_for=("goals_for","sum"),
        goals_against=("goals_against","sum")
    )
)

league_table["goal_diff"] = league_table["goals_for"] - league_table["goals_against"]
league_table["win_pct"] = league_table["wins"] / league_table["games_played"]

league_table = league_table.sort_values("win_pct", ascending=False)

league_table.head()
```

```python
# Add xG
home_xg = games[["home_team","home_xg","away_xg"]].copy()
home_xg.columns = ["team","xg_for","xg_against"]

away_xg = games[["away_team","away_xg","home_xg"]].copy()
away_xg.columns = ["team","xg_for","xg_against"]

xg_table = pd.concat([home_xg, away_xg]).groupby("team").sum()

league_table = league_table.join(xg_table)

league_table["xg_diff"] = league_table["xg_for"] - league_table["xg_against"]

# Power score (weighted)
league_table["power_score"] = (
    0.5 * league_table["xg_diff"] +
    0.3 * league_table["goal_diff"] +
    0.2 * league_table["win_pct"]
)

power_rankings = league_table.sort_values("power_score", ascending=False)

power_rankings.head(10)
```

```python
# Create feature differences
games["goal_diff"] = games["home_goals"] - games["away_goals"]
games["xg_diff"] = games["home_xg"] - games["away_xg"]
games["shot_diff"] = games["home_shots"] - games["away_shots"]

features = games[["xg_diff","shot_diff"]]
labels = games["home_win"]
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

X_train, X_test, y_train, y_test = train_test_split(
    features, labels, test_size=0.2, random_state=42
)

model = LogisticRegression()
model.fit(X_train, y_train)

probs = model.predict_proba(X_test)[:,1]

print("AUC:", roc_auc_score(y_test, probs))
```

```python
# ============================================================
# WHSDSC 2026 – Phase 1 (A–D) END-TO-END PIPELINE (1 cell)
# NO FILE LOADING HERE.
# Assumes you ALREADY have:
#   whl       = pd.read_csv("whl_2025.csv")
#   matchups  = pd.read_csv("WHSDSC_Rnd1_matchups.csv")
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, log_loss, brier_score_loss, accuracy_score, classification_report


# -----------------------------
# 0) Utilities: column detection
# -----------------------------
def _find_col(df, candidates, required=True):
    cols = set(df.columns)
    for c in candidates:
        if c in cols:
            return c
    if required:
        raise KeyError(
            f"None of these columns found: {candidates}\n"
            f"Available columns: {sorted(df.columns)}"
        )
    return None

def _infer_line_col(df, side, line_type="off"):
    """
    Infer offensive/defensive identifier columns.
    Works with common naming patterns; otherwise uses a fallback heuristic.
    """
    side = side.lower()
    lt = line_type.lower()

    if lt == "off":
        patterns = [
            f"{side}_offensive_line", f"{side}_off_line", f"{side}_oline", f"{side}_o_line",
            f"{side}_off_line_num", f"{side}_off_line_id", f"{side}_off_line_idx",
            f"{side}_line_off", f"{side}_line"
        ]
        fallback = [c for c in df.columns if side in c.lower() and "line" in c.lower() and "def" not in c.lower() and "pair" not in c.lower()]
    else:
        patterns = [
            f"{side}_defensive_pairing", f"{side}_def_pair", f"{side}_def_pairing",
            f"{side}_d_pair", f"{side}_d_pairing", f"{side}_defpair",
            f"{side}_pair_def", f"{side}_pair", f"{side}_def_line"
        ]
        fallback = [c for c in df.columns if side in c.lower() and ("pair" in c.lower() or ("def" in c.lower() and "line" in c.lower()))]

    for c in patterns:
        if c in df.columns:
            return c
    if fallback:
        return fallback[0]
    raise KeyError(f"Could not infer {side} {line_type} column. Available columns: {sorted(df.columns)}")

def _infer_toi_col(df, side):
    """
    Infer time-on-ice (TOI) column. Optional.
    """
    side = side.lower()
    candidates = [
        f"{side}_time_on_ice", f"{side}_toi", f"{side}_ice_time", f"{side}_time",
        f"{side}_seconds", f"{side}_sec"
    ]
    for c in candidates:
        if c in df.columns:
            return c
    fallback = [c for c in df.columns if side in c.lower() and ("toi" in c.lower() or "time_on_ice" in c.lower() or "ice_time" in c.lower())]
    if fallback:
        return fallback[0]
    return None

def _to_minutes(series):
    """
    Convert TOI to minutes if it looks like seconds.
    Rule: if median > 200 => likely seconds.
    """
    s = pd.to_numeric(series, errors="coerce")
    med = np.nanmedian(s.values)
    if np.isfinite(med) and med > 200:
        return s / 60.0
    return s


# --------------------------------
# 1) Phase 1a – Game-level dataset
# --------------------------------
game_id_col   = _find_col(whl, ["game_id"])
home_team_col = _find_col(whl, ["home_team"])
away_team_col = _find_col(whl, ["away_team"])

home_goals_col = _find_col(whl, ["home_goals"])
away_goals_col = _find_col(whl, ["away_goals"])
home_shots_col = _find_col(whl, ["home_shots"])
away_shots_col = _find_col(whl, ["away_shots"])
home_xg_col    = _find_col(whl, ["home_xg"])
away_xg_col    = _find_col(whl, ["away_xg"])

# Aggregate multiple rows per game -> 1 row per game
games = (
    whl
    .groupby([game_id_col, home_team_col, away_team_col], as_index=False)
    .agg(
        home_goals=(home_goals_col, "sum"),
        away_goals=(away_goals_col, "sum"),
        home_shots=(home_shots_col, "sum"),
        away_shots=(away_shots_col, "sum"),
        home_xg=(home_xg_col, "sum"),
        away_xg=(away_xg_col, "sum"),
    )
)
games["home_win"] = (games["home_goals"] > games["away_goals"]).astype(int)
games = games.loc[games["home_goals"] != games["away_goals"]].copy()

# Create sortable game index from game_id if possible
extracted = games[game_id_col].astype(str).str.extract(r"(\d+)")
games["game_num"] = pd.to_numeric(extracted[0], errors="coerce")
if games["game_num"].isna().all():
    games["game_num"] = np.arange(len(games))
games["game_num"] = games["game_num"].fillna(games["game_num"].median()).astype(int)
games = games.sort_values("game_num").reset_index(drop=True)

# League table
home_stats = games[[home_team_col, "home_goals", "away_goals", "home_xg", "away_xg", "home_win"]].copy()
home_stats.columns = ["team","goals_for","goals_against","xg_for","xg_against","win"]

away_stats = games[[away_team_col, "away_goals", "home_goals", "away_xg", "home_xg", "home_win"]].copy()
away_stats.columns = ["team","goals_for","goals_against","xg_for","xg_against","home_win"]
away_stats["win"] = 1 - away_stats["home_win"]
away_stats = away_stats.drop(columns=["home_win"])

team_games = pd.concat([home_stats, away_stats], ignore_index=True)

league_table = (
    team_games
    .groupby("team", as_index=True)
    .agg(
        games_played=("win","count"),
        wins=("win","sum"),
        goals_for=("goals_for","sum"),
        goals_against=("goals_against","sum"),
        xg_for=("xg_for","sum"),
        xg_against=("xg_against","sum"),
    )
)
league_table["losses"] = league_table["games_played"] - league_table["wins"]
league_table["goal_diff"] = league_table["goals_for"] - league_table["goals_against"]
league_table["xg_diff"] = league_table["xg_for"] - league_table["xg_against"]
league_table["win_pct"] = league_table["wins"] / league_table["games_played"]

# Power score: emphasizes underlying strength (xG), then results
league_table["power_score"] = (
    0.55 * league_table["xg_diff"] +
    0.30 * league_table["goal_diff"] +
    0.15 * league_table["win_pct"] * league_table["games_played"]
)

power_rankings = league_table.sort_values("power_score", ascending=False).reset_index()
power_rankings["power_rank"] = np.arange(1, len(power_rankings)+1)

print("=== PHASE 1a: Power Rankings (Top 10) ===")
display(power_rankings.head(10))


# -------------------------------------------------------
# 2) Phase 1a – Win probability model (Elo + priors)
#     (leakage-safe: only prior info used per game)
# -------------------------------------------------------
def build_sequential_features(games_df, base_elo=1500.0, k=20.0, home_adv=65.0):
    teams = set(games_df[home_team_col]).union(set(games_df[away_team_col]))
    elo = {t: base_elo for t in teams}
    stats = defaultdict(lambda: {"gp":0,"wins":0,"gf":0.0,"ga":0.0,"xgf":0.0,"xga":0.0,"sf":0.0,"sa":0.0})

    rows = []
    for _, r in games_df.sort_values("game_num").iterrows():
        ht, at = r[home_team_col], r[away_team_col]
        st_h, st_a = stats[ht], stats[at]

        def rate(num, den, default):
            return num/den if den > 0 else default

        elo_home = elo.get(ht, base_elo)
        elo_away = elo.get(at, base_elo)
        elo_diff = (elo_home + home_adv) - elo_away
        exp_home = 1.0 / (1.0 + 10 ** (-(elo_diff / 400.0)))

        home_gp, away_gp = st_h["gp"], st_a["gp"]

        feats = {
            "home_elo": elo_home,
            "away_elo": elo_away,
            "elo_diff": elo_diff,
            "elo_exp_home": exp_home,

            "home_gp": home_gp,
            "away_gp": away_gp,

            "home_win_pct": rate(st_h["wins"], home_gp, 0.5),
            "away_win_pct": rate(st_a["wins"], away_gp, 0.5),

            "home_goal_diff_avg": rate(st_h["gf"] - st_h["ga"], home_gp, 0.0),
            "away_goal_diff_avg": rate(st_a["gf"] - st_a["ga"], away_gp, 0.0),

            "home_xg_diff_avg": rate(st_h["xgf"] - st_h["xga"], home_gp, 0.0),
            "away_xg_diff_avg": rate(st_a["xgf"] - st_a["xga"], away_gp, 0.0),

            "home_shot_diff_avg": rate(st_h["sf"] - st_h["sa"], home_gp, 0.0),
            "away_shot_diff_avg": rate(st_a["sf"] - st_a["sa"], away_gp, 0.0),
        }

        y = int(r["home_goals"] > r["away_goals"])
        feats["label_home_win"] = y
        rows.append(feats)

        # Update Elo
        elo[ht] = elo_home + k * (y - exp_home)
        elo[at] = elo_away + k * ((1 - y) - (1 - exp_home))

        # Update stats
        st_h["gp"] += 1; st_a["gp"] += 1
        st_h["wins"] += y; st_a["wins"] += (1 - y)

        st_h["gf"] += r["home_goals"]; st_h["ga"] += r["away_goals"]
        st_a["gf"] += r["away_goals"]; st_a["ga"] += r["home_goals"]

        st_h["xgf"] += r["home_xg"]; st_h["xga"] += r["away_xg"]
        st_a["xgf"] += r["away_xg"]; st_a["xga"] += r["home_xg"]

        st_h["sf"] += r["home_shots"]; st_h["sa"] += r["away_shots"]
        st_a["sf"] += r["away_shots"]; st_a["sa"] += r["home_shots"]

    feat_df = pd.DataFrame(rows)
    return feat_df, elo, stats

feat_df, final_elo, final_stats = build_sequential_features(games)

# Train/test split by time (last 20% as test)
df = feat_df.copy()

# IMPORTANT FIX: do NOT include game_num as a model feature (we removed it entirely)
X = df.drop(columns=["label_home_win"])
y = df["label_home_win"].astype(int).values

split = int(len(df) * 0.8)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y[:split], y[split:]

model = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(max_iter=2000))
])
model.fit(X_train, y_train)

p_test = model.predict_proba(X_test)[:, 1]
pred_test = (p_test >= 0.5).astype(int)

print("\n=== PHASE 1a: Model performance (time-based split) ===")
print(f"AUC:      {roc_auc_score(y_test, p_test):.4f}")
print(f"LogLoss:  {log_loss(y_test, p_test):.4f}")
print(f"Brier:    {brier_score_loss(y_test, p_test):.4f}")
print(f"Acc@0.5:  {accuracy_score(y_test, pred_test):.4f}")
print("\nClassification report (@0.5):")
print(classification_report(y_test, pred_test, digits=4))


# -----------------------------------------
# 3) Phase 1a – Predict Round 1 matchups
# -----------------------------------------
m_home_team = _find_col(matchups, ["home_team"])
m_away_team = _find_col(matchups, ["away_team"])

def matchup_features(matchups_df, final_elo, final_stats, base_elo=1500.0, home_adv=65.0):
    rows = []
    for _, r in matchups_df.iterrows():
        ht, at = r[m_home_team], r[m_away_team]

        elo_home = final_elo.get(ht, base_elo)
        elo_away = final_elo.get(at, base_elo)
        elo_diff = (elo_home + home_adv) - elo_away
        exp_home = 1.0 / (1.0 + 10 ** (-(elo_diff / 400.0)))

        st_h = final_stats.get(ht, {"gp":0,"wins":0,"gf":0,"ga":0,"xgf":0,"xga":0,"sf":0,"sa":0})
        st_a = final_stats.get(at, {"gp":0,"wins":0,"gf":0,"ga":0,"xgf":0,"xga":0,"sf":0,"sa":0})

        def rate(num, den, default):
            return num/den if den > 0 else default

        home_gp, away_gp = st_h["gp"], st_a["gp"]

        rows.append({
            "home_elo": elo_home,
            "away_elo": elo_away,
            "elo_diff": elo_diff,
            "elo_exp_home": exp_home,

            "home_gp": home_gp,
            "away_gp": away_gp,

            "home_win_pct": rate(st_h["wins"], home_gp, 0.5),
            "away_win_pct": rate(st_a["wins"], away_gp, 0.5),

            "home_goal_diff_avg": rate(st_h["gf"] - st_h["ga"], home_gp, 0.0),
            "away_goal_diff_avg": rate(st_a["gf"] - st_a["ga"], away_gp, 0.0),

            "home_xg_diff_avg": rate(st_h["xgf"] - st_h["xga"], home_gp, 0.0),
            "away_xg_diff_avg": rate(st_a["xgf"] - st_a["xga"], away_gp, 0.0),

            "home_shot_diff_avg": rate(st_h["sf"] - st_h["sa"], home_gp, 0.0),
            "away_shot_diff_avg": rate(st_a["sf"] - st_a["sa"], away_gp, 0.0),
        })
    return pd.DataFrame(rows)

X_match = matchup_features(matchups, final_elo, final_stats)

# Align columns exactly to training features
X_match = X_match[X.columns]

matchups_out = matchups.copy()
matchups_out["home_win_probability"] = model.predict_proba(X_match)[:, 1]
matchups_out["away_win_probability"] = 1.0 - matchups_out["home_win_probability"]
matchups_out["predicted_winner"] = np.where(
    matchups_out["home_win_probability"] >= 0.5,
    matchups_out[m_home_team],
    matchups_out[m_away_team]
)

print("\n=== PHASE 1a: Round 1 predictions ===")
display(matchups_out)

matchups_out.to_csv("Phase1a_Round1_Predictions.csv", index=False)
power_rankings.to_csv("Phase1a_PowerRankings.csv", index=False)
print("\nSaved: Phase1a_Round1_Predictions.csv, Phase1a_PowerRankings.csv")


# -----------------------------------------
# 4) Phase 1b – Offensive line disparity (xG per minute)
# -----------------------------------------
home_off_line_col = _infer_line_col(whl, "home", "off")
away_off_line_col = _infer_line_col(whl, "away", "off")

home_toi_col = _infer_toi_col(whl, "home")
away_toi_col = _infer_toi_col(whl, "away")

# Build long-format line table from home + away
home_lines = whl[[home_team_col, home_off_line_col, home_xg_col] + ([home_toi_col] if home_toi_col else [])].copy()
home_lines.columns = ["team","offensive_line","xg"] + (["toi"] if home_toi_col else [])
if home_toi_col:
    home_lines["toi_min"] = _to_minutes(home_lines["toi"])
else:
    home_lines["toi_min"] = 1.0  # fallback

away_lines = whl[[away_team_col, away_off_line_col, away_xg_col] + ([away_toi_col] if away_toi_col else [])].copy()
away_lines.columns = ["team","offensive_line","xg"] + (["toi"] if away_toi_col else [])
if away_toi_col:
    away_lines["toi_min"] = _to_minutes(away_lines["toi"])
else:
    away_lines["toi_min"] = 1.0  # fallback

line_long = pd.concat(
    [
        home_lines[["team","offensive_line","xg","toi_min"]],
        away_lines[["team","offensive_line","xg","toi_min"]],
    ],
    ignore_index=True
)

line_stats = (
    line_long
    .groupby(["team","offensive_line"], as_index=False)
    .agg(xg=("xg","sum"), toi_min=("toi_min","sum"))
)

line_stats["xg_per_min"] = line_stats["xg"] / line_stats["toi_min"].replace({0: np.nan})
line_stats = line_stats.dropna(subset=["xg_per_min"])

line_pivot = line_stats.pivot(index="team", columns="offensive_line", values="xg_per_min")

# Identify which columns correspond to line 1 and line 2
cols = list(line_pivot.columns)

def _pick_line_col(target):
    # exact numeric
    for c in cols:
        try:
            if int(c) == target:
                return c
        except:
            pass
    # string contains
    key = "first" if target == 1 else "second"
    for c in cols:
        if key in str(c).lower():
            return c
    # fallback: numeric-sort
    def _to_num(x):
        try: return float(x)
        except: return np.nan
    nums = [(c, _to_num(c)) for c in cols]
    nums_sorted = sorted(nums, key=lambda t: (np.isnan(t[1]), t[1]))
    if len(nums_sorted) >= target:
        return nums_sorted[target-1][0]
    raise KeyError(f"Could not find line {target} column among: {cols}")

col1 = _pick_line_col(1)
col2 = _pick_line_col(2)

disparity = line_pivot.copy()
disparity["disparity_ratio"] = disparity[col1] / disparity[col2]
disparity_rank = disparity.sort_values("disparity_ratio", ascending=False).reset_index()

top10_disparity = disparity_rank[["team","disparity_ratio"]].head(10)

print("\n=== PHASE 1b: Top 10 Offensive Line Disparity Teams ===")
display(top10_disparity)

top10_disparity.to_csv("Phase1b_Top10_LineDisparity.csv", index=False)
print("Saved: Phase1b_Top10_LineDisparity.csv")


# -----------------------------------------
# 5) Phase 1c – Visualization: disparity vs strength
# -----------------------------------------
viz = disparity_rank.merge(power_rankings[["team","power_score","power_rank"]], on="team", how="left")

plt.figure(figsize=(9,6))
plt.scatter(viz["disparity_ratio"], viz["power_score"])
plt.xlabel("Offensive Line Disparity Ratio (Line1 xG/min ÷ Line2 xG/min)")
plt.ylabel("Team Strength (Power Score)")
plt.title("Phase 1c: Offensive Line Disparity vs Team Strength")
plt.axvline(viz["disparity_ratio"].median(), linestyle="--")
plt.text(viz["disparity_ratio"].median(), plt.ylim()[1]*0.95, "Median disparity", rotation=90, va="top")
plt.tight_layout()
plt.savefig("Phase1c_Disparity_vs_Strength.png", dpi=300)
plt.show()
print("Saved: Phase1c_Disparity_vs_Strength.png")


# -----------------------------------------
# 6) Phase 1d – Paste-ready methodology text
# -----------------------------------------
phase1d = {}

phase1d["1) Process (~50 words)"] = (
    "We cleaned the season data by aggregating the multiple line-matchup rows into one row per game, "
    "summing goals, shots, and expected goals (xG) for home and away teams. We removed any tied games "
    "to keep a binary win/loss outcome. We created a time-ordered game index from game_id to support "
    "leakage-safe feature engineering and evaluation."
)

phase1d["2) Additional variables (~25 words)"] = (
    "We engineered pre-game features including Elo ratings, prior win percentage, and prior per-game "
    "goal/xG/shot differentials for both home and away teams."
)

phase1d["3) Tools & purposes (~50 words)"] = (
    "We used Python in Google Colab with pandas/numpy for cleaning, aggregation, and feature engineering. "
    "We used scikit-learn to train a probabilistic logistic regression model and compute evaluation metrics. "
    "We used matplotlib to create a clear scatter plot and export it as a PNG for submission."
)

phase1d["4) Statistical methods (~100 words)"] = (
    "To predict win probabilities, we built a leakage-safe sequential pipeline. For each game in time order, "
    "we computed team strength features using only information available prior to that game: Elo ratings (updated after each game "
    "using expected-score formulas and a home-advantage constant) and rolling prior performance summaries (win%, and per-game differentials "
    "in goals, xG, and shots). We trained a logistic regression model on these pre-game features and evaluated on the most recent 20% of games "
    "to mimic forecasting future matchups. The model outputs calibrated win probabilities for the home team."
)

phase1d["5) 1a Predictions (~50 words)"] = (
    "Power rankings were based on season-long indicators emphasizing underlying performance (xG differential, then goal differential and win rate). "
    "Round 1 win probabilities were produced by computing each team’s pre-match features (final Elo and season priors) and applying the trained model "
    "to estimate the probability the home team wins."
)

phase1d["6) 1b Disparity (~50 words)"] = (
    "We quantified offensive line quality using xG per minute for each team’s first and second offensive lines, combining home and away appearances. "
    "We summed xG and total time on ice (TOI) by team and line, computed xG/TOI, and formed a disparity ratio as (Line 1 xG/min) ÷ (Line 2 xG/min). "
    "Teams were ranked by this ratio to identify the largest disparities."
)

phase1d["7) 1c Visualization (~50 words)"] = (
    "We used a scatter plot with offensive line disparity on the x-axis and team strength (power score) on the y-axis, with clear labels and a descriptive title. "
    "This design makes it easy to see whether more evenly matched lines (lower disparity) are associated with stronger overall teams."
)

phase1d["8) Model performance (~50 words)"] = (
    "We assessed performance with a time-based holdout set (last 20% of games). We reported AUC to measure ranking quality, log loss and Brier score to measure "
    "probabilistic accuracy and calibration, and accuracy at a 0.5 threshold. We also reviewed the classification report for class-level precision and recall."
)

phase1d["9) Generative AI use (~50 words)"] = (
    "We used generative AI to accelerate coding and documentation: drafting aggregation steps, proposing leakage-safe features (Elo + prior performance), and producing "
    "a reproducible methodology write-up. We validated the approach by checking column mappings, confirming one-row-per-game aggregation, and verifying outputs through "
    "holdout evaluation metrics and sanity checks."
)

print("\n=== PHASE 1d: Paste-ready methodology responses ===")
for k, v in phase1d.items():
    print(f"\n{k}\n{v}")

with open("Phase1d_Methodology_Text.txt", "w", encoding="utf-8") as f:
    for k, v in phase1d.items():
        f.write(k + "\n" + v + "\n\n")
print("\nSaved: Phase1d_Methodology_Text.txt")

print("\nDONE ✅ Outputs created:")
print(" - Phase1a_PowerRankings.csv")
print(" - Phase1a_Round1_Predictions.csv")
print(" - Phase1b_Top10_LineDisparity.csv")
print(" - Phase1c_Disparity_vs_Strength.png")
print(" - Phase1d_Methodology_Text.txt")
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, log_loss, brier_score_loss, accuracy_score, classification_report

# ----------------------------
# Build team-strength features
# ----------------------------
g = games.copy()

# One-hot encode home teams and away teams separately
home_dummies = pd.get_dummies(g["home_team"], prefix="H")
away_dummies = pd.get_dummies(g["away_team"], prefix="A")

X_bt = pd.concat([home_dummies, away_dummies], axis=1)

# Make sure columns align as: +1 for home team, -1 for away team (Bradley-Terry style)
# Convert into shared team columns by subtracting away indicators from home indicators
teams = sorted(set(g["home_team"]).union(set(g["away_team"])))

X = pd.DataFrame(0, index=g.index, columns=[f"T_{t}" for t in teams])
for t in teams:
    X[f"T_{t}"] = (g["home_team"] == t).astype(int) - (g["away_team"] == t).astype(int)

y = g["home_win"].astype(int).values

# Time-based split (like you used)
g = g.sort_values("game_num").reset_index(drop=True)
X = X.loc[g.index]
y = g["home_win"].astype(int).values

split = int(len(g) * 0.8)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y[:split], y[split:]

# Regularized logistic regression (prevents overfitting)
bt_model = LogisticRegression(max_iter=5000, C=1.0)
bt_model.fit(X_train, y_train)

p_test = bt_model.predict_proba(X_test)[:, 1]
pred_test = (p_test >= 0.5).astype(int)

print("=== Bradley–Terry Team-Strength Model Performance ===")
print(f"AUC:      {roc_auc_score(y_test, p_test):.4f}")
print(f"LogLoss:  {log_loss(y_test, p_test):.4f}")
print(f"Brier:    {brier_score_loss(y_test, p_test):.4f}")
print(f"Acc@0.5:  {accuracy_score(y_test, pred_test):.4f}")
print("\nClassification report (@0.5):")
print(classification_report(y_test, pred_test, digits=4))

# ----------------------------
# Power rankings from weights
# ----------------------------
strength = pd.Series(bt_model.coef_[0], index=X.columns)
power_rankings_bt = (
    strength.sort_values(ascending=False)
    .reset_index()
)
power_rankings_bt.columns = ["team_col", "strength"]
power_rankings_bt["team"] = power_rankings_bt["team_col"].str.replace("^T_", "", regex=True)
power_rankings_bt["power_rank"] = np.arange(1, len(power_rankings_bt)+1)

print("\n=== Power Rankings from Team Strength (Top 10) ===")
display(power_rankings_bt[["team","strength","power_rank"]].head(10))

power_rankings_bt[["team","strength","power_rank"]].to_csv("Phase1a_PowerRankings_BT.csv", index=False)

# ----------------------------
# Predict Round 1 matchups
# ----------------------------
def matchup_matrix(matchups_df, teams):
    Xm = pd.DataFrame(0, index=matchups_df.index, columns=[f"T_{t}" for t in teams])
    for t in teams:
        Xm[f"T_{t}"] = (matchups_df["home_team"] == t).astype(int) - (matchups_df["away_team"] == t).astype(int)
    return Xm

X_match = matchup_matrix(matchups, teams)
matchups_bt = matchups.copy()
matchups_bt["home_win_probability"] = bt_model.predict_proba(X_match)[:, 1]
matchups_bt["away_win_probability"] = 1 - matchups_bt["home_win_probability"]
matchups_bt["predicted_winner"] = np.where(matchups_bt["home_win_probability"] >= 0.5,
                                           matchups_bt["home_team"], matchups_bt["away_team"])

print("\n=== Round 1 Predictions (BT Model) ===")
display(matchups_bt)

matchups_bt.to_csv("Phase1a_Round1_Predictions_BT.csv", index=False)
print("\nSaved: Phase1a_PowerRankings_BT.csv, Phase1a_Round1_Predictions_BT.csv")
```
